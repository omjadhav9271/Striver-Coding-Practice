03 Using only methods

```python
def find_max(arr):
    if not arr:
        raise ValueError("Array must not be empty")
    max_val = arr[0]
    for x in arr[1:]:
        if x > max_val:
            max_val = x
    return max_val

def main():
    raw = input("Enter numbers separated by spaces: ")
    arr = list(map(int, raw.split()))
    print("Max value:", find_max(arr))

if __name__ == "__main__":
    main()
```

04 Using Class

```python
class ArrayMaxFinder:
    def __init__(self, arr):
        self.arr = arr

    def compute_max(self):
        if not self.arr:
            raise ValueError("Array must not be empty")

        max_val = self.arr[0]
        for x in self.arr[1:]:
            if x > max_val:
                max_val = x
        return max_val

def main():
    raw = input("Enter numbers separated by spaces: ")
    arr = list(map(int, raw.split()))
    finder = ArrayMaxFinder(arr)
    print("Max value:", finder.compute_max())

if __name__ == "__main__":
    main()
```

05 Extract pure functions

```python
# arr_utils.py
def find_max(arr):
    """Return max element; raises ValueError if empty."""
    if not arr:
        raise ValueError("Array must not be empty")
    max_val = arr[0]
    for x in arr[1:]:
        if x > max_val:
            max_val = x
    return max_val
```

```python
# cli.py
from arr_utils import find_max

def main():
    raw = input("Enter numbers: ")
    arr = list(map(int, raw.split()))
    print("Max value:", find_max(arr))

if __name__ == "__main__":
    main()
```
